```python
import pandas as pd

# Get only the duplicated tx hashes (the ones with >1 mint)
dupe_txs = mint.groupby('evt_tx_hash').filter(lambda x: len(x) > 1)['evt_tx_hash'].unique()

# Work through them
unresolved = []
resolved_matches = []

for tx in dupe_txs:
    mints = mint[mint['evt_tx_hash'] == tx]
    opens = open_events[open_events['evt_tx_hash'] == tx]
    
    for _, mint_row in mints.iterrows():
        # match on liquidity_minted == liquidity_added
        candidates = opens[opens['liquidity_added'] == mint_row['liquidity_minted']]
        
        if len(candidates) == 1:
            resolved_matches.append({
                'evt_tx_hash': tx,
                'tokenId': candidates.iloc[0]['tokenId'],
                'tickLower': mint_row['tickLower'],
                'tickUpper': mint_row['tickUpper'],
            })
        else:
            unresolved.append(tx)

print(f"Resolved: {len(resolved_matches)}")
print(f"Unresolved (ambiguous or no match): {len(set(unresolved))}")
```

```python
no_match = []
multi_match = []

for tx in dupe_txs:
    mints = mint[mint['evt_tx_hash'] == tx]
    opens = open_events[open_events['evt_tx_hash'] == tx]
    
    for _, mint_row in mints.iterrows():
        candidates = opens[opens['liquidity_added'] == mint_row['liquidity_minted']]
        if len(candidates) == 0:
            no_match.append(tx)
        elif len(candidates) > 1:
            multi_match.append(tx)

print(f"No match: {len(set(no_match))}")
print(f"Multiple matches: {len(set(multi_match))}")
```
